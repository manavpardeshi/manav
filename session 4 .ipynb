```python
import json

songs = [
    {"song": "Blinding Lights", "artist": "The Weeknd", "streams": 350000000},
    {"song": "Shape of You", "artist": "Ed Sheeran", "streams": 300000000},
    {"song": "As It Was", "artist": "Harry Styles", "streams": 250000000},
    {"song": "Bad Guy", "artist": "Billie Eilish", "streams": 220000000},
    {"song": "Stay", "artist": "The Kid LAROI & Justin Bieber", "streams": 200000000}
]

# Create the JSON file
with open("trending_songs.json", "w", encoding="utf-8") as f:
    json.dump(songs, f, indent=4)

print("trending_songs.json created successfully!")
```

```python
import pandas as pd

# Load JSON into DataFrame
df = pd.read_json("trending_songs.json")

# Display the DataFrame
print(df)

# Display column names and data types
df.info()
```
